```python
import pandas as pd
import sys
import os

sys.path.append(os.path.abspath(".."))

from scripts.scrap import get_info, scrap, extract_columns
from scripts.preprocess import preprocess_reviews
```

```python
banks = {
    "Commercial Bank of Ethiopia": "com.combanketh.mobilebanking",
    "Bank of Abyssinia": "com.boa.boaMobileBanking",
    "Dashen Bank": "com.dashen.dashensuperapp"
}
```

## Web scraping

```python
raw_data = {}

for name, app_id in banks.items():
    result = scrap(app_id)
    df = extract_columns(result, name)
    raw_data[name] = df
```

```python
for name, df in raw_data.items():
    print(name, len(df))
    display(df.head())
```

## Cleaning data

```python
clean_data = {}

for name, df in raw_data.items():
    clean_data[name] = preprocess_reviews(df)
```

```python
for name, df in clean_data.items():
    print(name)
    display(df.head())
```

## Saving files

```python
os.makedirs("../data/processed", exist_ok=True)

for name, df in clean_data.items():
    file_name = name.lower().replace(" ", "_")
    path = f"../data/processed/{file_name}_reviews_clean.csv"
    df.to_csv(path, index=False)
    print("Saved:", path)
```
