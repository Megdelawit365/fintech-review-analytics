# Cleaning data

```python
df_copy = raw_data.copy()
```

### Handling Missing Values
No missing values were found in all the datasets so missing values need not be handled

### Removing Duplicate Values

There are no reviews with the same id, so duplicate value handling is also not needed

### Normalizing Dates

```python
for name, df in df_copy.items():
    print("="*50)
    print(name)
    print(" ")

    print("Before normalization:")
    print(df['date'].head(3).to_string())
    print(f"dtype: {df['date'].dtype}")

    df['date'] = pd.to_datetime(df['date']).dt.strftime('%Y-%m-%d')

    print("\nAfter normalization:")
    print(df['date'].head(3).to_string())
    print(f"dtype: {df['date'].dtype}")

    print(f"\nDate range: {df['date'].min()} to {df['date'].max()}")
    print("")


```

### Cleaning review text

```python
def clean_text(text):
        if pd.isna(text):
            return ''
        text = str(text)
        text = re.sub(r'\s+', ' ', text)  
        text = text.strip()               
        return text
for name, df in df_copy.items():
    print("="*50)
    print(name)

    df['review'] = df['review'].apply(clean_text)

    before = len(df)
    df = df[df['review'].str.len() > 0]
    removed = before - len(df)
    print(f"\nRemoved {removed} reviews that were empty after cleaning")
    print("")


```

### Validating ratings

```python
for name, df in df_copy.items():
    print("="*50)
    print(name)

    invalid_ratings = df[(df['rating'] < 1) | (df['rating'] > 5)]
    print(f"Invalid ratings (outside 1–5): {len(invalid_ratings)}")

    df = df[(df['rating'] >= 1) & (df['rating'] <= 5)]

    df['rating'] = df['rating'].astype(int)

    print(f"Remaining: {len(df)} reviews")
    print(f"Rating dtype: {df['rating'].dtype}")
    print("")
```

# Saving clean data

```python
df_clean = {}
for name, df in df_copy.items():
    df_clean[name] = df[['review', 'rating', 'date', 'bank', 'source']].copy()
    df_clean[name] = df_clean[name].sort_values('date', ascending=False).reset_index(drop=True)
for name, df in df_clean.items():
    print(name)
    print(" ")
    print("="*100)
    display(df.head(5))
```

```python
os.makedirs('../data/processed', exist_ok=True)
for name, df in df_clean.items():
    name = re.sub(r'\s+', '_', name) 
    output_path = f"../data/processed/{name.lower()}_reviews_clean.csv"
    df.to_csv(output_path, index=False)

    print(f"Saved to: {output_path}")
```
