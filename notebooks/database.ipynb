```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
DATABASE_URL = "postgresql://postgres:1234@localhost:5432/bank_reviews"
engine = create_engine(DATABASE_URL)

df = pd.read_csv("../data/processed/final_reviews.csv")
df.head()
```

```python
banks = pd.read_sql("SELECT * FROM banks", engine)
banks
```

```python
reviews = pd.read_sql("SELECT * FROM reviews LIMIT 10", engine)
reviews
```

```python
query = """
SELECT b.bank_name, COUNT(*) AS total_reviews
FROM reviews r
JOIN banks b
ON r.bank_id = b.bank_id
GROUP BY b.bank_name
"""

pd.read_sql(query, engine)
```
